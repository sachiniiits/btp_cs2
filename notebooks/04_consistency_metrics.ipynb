# 04 · Role-consistency metrics (the novel contribution)
Per player-half: **modal role share**, normalized role entropy, switch rate.
Plus the **rolling** (prior-rounds-only) version that feeds the outcome model —
computed strictly from rounds *before* round *t* in the same half, so the
regression cannot leak outcome information backwards.

```python
labeled = roles.load_labeled()
mlv = cons.match_level(labeled)
print(mlv[['modal_share','entropy_norm','switch_rate']].describe().round(3))
viz.consistency_hist(mlv);
```

## Face validity: most / least consistent player-halves
You should recognize the archetypes — dedicated AWPers and anchors near the top.

```python
cols = ['player_name','team_clan','side','n_rounds','modal_share']
print('MOST consistent:');  print(mlv.nlargest(10,'modal_share')[cols].to_string(index=False))
print('\nLEAST consistent:'); print(mlv.nsmallest(10,'modal_share')[cols].to_string(index=False))
```

```python
rolled = cons.rolling(labeled)
team_round = cons.team_round_table(rolled)
print('team-round rows:', len(team_round),
      '| usable (non-NaN):', team_round.cons_mean.notna().sum())
team_round.head()
```
